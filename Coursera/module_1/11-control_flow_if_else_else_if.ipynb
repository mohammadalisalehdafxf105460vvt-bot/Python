```python
# Example for Conditional statments using if , else , and elif.
bill_total = float(input("What is the total ?"))
discount1 = 10
if bill_total > 100:
    print("The bill is grater than 100")
    bill_total -= discount1
else:
    print("the bills totla", bill_total , "is less than 100 no need for discount")
print("total bill is :- "+str(bill_total))
```
