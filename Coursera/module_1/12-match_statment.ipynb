# We will learn how to use match case statments 
# which it's like switch in C++ 
## To type the or operator we will put | 
## let's see we want to write else statment; to do  that we will ad case _:

```python
https_statues = int(input("what is the number of the statues ? "))
match https_statues:
    case 200:
        print(https_statues,"Succsess")
    case 400:
         print(https_statues,"Bad request")
    case 404:
         print(https_statues,"Not Found")
    # will add also 501 so we will use |  
    case 500 | 501:
         print(https_statues,"Server Error")
    # Now we want the rest of the conditions be like a catch
    # by printing unknown enrty 
    # we will writ case _:

    case _:
        print("Bad Entry!")
```
